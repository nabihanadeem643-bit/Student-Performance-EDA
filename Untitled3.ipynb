```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import zipfile
import io
import urllib.request
```

2. Univariate EDA

2.1 Distribution of Final Grades (G3)

A histogram is used to understand the distribution of students' final mathematics grades. It helps identify the marks range in which most students fall.

```python

plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="G3", bins=15, kde=True)

plt.title("Distribution of Final Mathematics Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```

Interpretation:

 The histogram shows the distribution of students' final mathematics grades. It helps us understand how students performed in the examination. The graph also shows the range of marks and whether most students received low, medium, or high grades.

```python

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="studytime")

plt.title("Distribution of Students by Study Time")
plt.xlabel("Study Time")
plt.ylabel("Number of Students")

plt.show()
```

Interpretation:

The bar chart shows the number of students in each study-time category. It helps us understand how much time students spend studying. The categories represent different levels of weekly study time.

```python

plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="absences", bins=20, kde=True)

plt.title("Distribution of Student Absences")
plt.xlabel("Number of Absences")
plt.ylabel("Number of Students")

plt.show()
```

Interpretation:

The histogram shows the distribution of student absences. It helps identify how many students have low or high numbers of absences. Students with unusually high absences may represent potential outliers.

```python
plt.figure(figsize=(6, 5))

sns.countplot(data=df, x="sex")

plt.title("Distribution of Students by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Students")

plt.show()
```

Interpretation:

The bar chart compares the number of male and female students in the dataset. It helps us understand the gender distribution and identify whether both groups are represented equally.

Boxplot of Final Mathematics Grades (G3)

A boxplot is used to understand the spread of final grades, the median, quartiles, and possible outliers.

```python

plt.figure(figsize=(8, 4))

sns.boxplot(data=df, x="G3")

plt.title("Boxplot of Final Mathematics Grades")
plt.xlabel("Final Grade (G3)")

plt.show()
```

Interpretation:

The boxplot shows the median, spread, and quartiles of students' final mathematics grades. Points outside the whiskers may indicate unusually low or high grades. These observations should be examined before deciding whether they are genuine outliers.

Distribution of Target Variable (G3)

The target variable is G3, which represents students' final mathematics grades. Its distribution helps us understand students' academic performance and the variation in their final marks.

```python
print("Final Grade Distribution:")
print(df["G3"].value_counts().sort_index())

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="G3")

plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.show()
```

Interpretation:

The bar chart shows the number of students who received each final grade. It helps identify the most common grades and shows how student performance varies across the dataset. The target variable is numerical, so its distribution is not described as balanced or imbalanced in exactly the same way as a classification target.

Relationship Between Study Time and Final Grade

This analysis explores the relationship between students' weekly study time and their final mathematics grades. It helps us investigate whether students with different study-time levels show differences in academic performance.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="studytime", y="G3")

plt.title("Study Time vs Final Grade")
plt.xlabel("Study Time Category")
plt.ylabel("Final Grade (G3)")

plt.show()
```

Interpretation:

The boxplot compares final mathematics grades across different study-time categories. It helps us examine whether students who spend different amounts of time studying have different grade distributions. The results should be interpreted from the observed medians and spread of each group.

Relationship Between Gender and Final Grade

This analysis compares the final mathematics grades of male and female students to explore possible differences in academic performance between the two groups.

```python

plt.figure(figsize=(7, 5))

sns.boxplot(data=df, x="sex", y="G3")

plt.title("Gender vs Final Mathematics Grade")
plt.xlabel("Gender")
plt.ylabel("Final Grade (G3)")

plt.show()
```

Interpretation:

The boxplot compares the distribution of final mathematics grades for male and female students. It helps identify differences in median grades, variation, and possible outliers between the two groups. These differences describe the dataset and do not necessarily mean that gender causes differences in academic performance.

Relationship Between Absences and Final Grade

This analysis examines the relationship between the number of student absences and final mathematics grades. It helps us explore whether students with more absences tend to have different academic outcomes.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="absences", y="G3")

plt.title("Absences vs Final Mathematics Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade (G3)")

plt.show()
```

Interpretation:

The scatter plot shows the relationship between student absences and final mathematics grades. Each point represents a student. The pattern of points helps us examine whether higher absences are associated with lower grades or whether no clear relationship is visible.

Correlation Analysis

Correlation analysis is used to examine the relationships between numerical variables in the Student Performance dataset. A correlation heatmap helps identify positive, negative, and weak linear relationships between these variables.

```python
numeric_df = df.select_dtypes(include="number")

plt.figure(figsize=(14, 10))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.show()
```

Interpretation:

The correlation heatmap shows the strength and direction of linear relationships between numerical variables. Positive values indicate that two variables tend to increase together, while negative values indicate that one tends to decrease as the other increases. Values close to zero indicate weak linear relationships. The correlations between G1, G2, and G3 are particularly important because they show how previous grades relate to final performance.

 6. Multivariate EDA

Multivariate EDA examines three or more variables together to identify combined patterns and relationships. In this analysis, study time, gender, and final mathematics grades are explored together to understand differences in academic performance across student groups.

```python

sns.catplot(
    data=df,
    x="studytime",
    y="G3",
    hue="sex",
    kind="box",
    height=5,
    aspect=1.4
)

plt.title("Study Time, Gender, and Final Grade")
plt.xlabel("Study Time Category")
plt.ylabel("Final Grade (G3)")

plt.show()
```

Interpretation:

This graph compares final mathematics grades across study-time categories for male and female students. It helps identify whether grade distributions differ by study time and gender. The combined patterns may reveal differences between groups, but they do not establish cause and effect.
